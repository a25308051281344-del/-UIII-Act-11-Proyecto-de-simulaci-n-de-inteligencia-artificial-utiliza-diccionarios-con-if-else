```python
print("===================================")
print("      IA CHEF MICHELIN")
print("===================================")

while True:

    print("\nMENÚ PRINCIPAL")
    print("1. Crear chef Michelin")
    print("2. Diseñar menú gourmet")
    print("3. Organizar brigada de cocina")
    print("4. Salir")

    opcion = input("Selecciona una opción: ")

    # OPCIÓN 1
    if opcion == "1":

        print("\n--- CHEF MICHELIN ---")

        estrellas = input("¿Cuántas estrellas Michelin tiene? ")

        if estrellas == "1":
            print("IA: Chef reconocido a nivel local.")

        elif estrellas == "2":
            print("IA: Chef con cocina de alta calidad.")

        elif estrellas == "3":
            print("IA: Chef de clase mundial.")

        else:
            print("IA: Chef en proceso de reconocimiento.")

        especialidad = input("¿Especialidad culinaria? ")

        if especialidad == "italiana":
            print("IA: Preparando recetas gourmet italianas.")

        elif especialidad == "mexicana":
            print("IA: Creando platillos tradicionales modernos.")

        elif especialidad == "francesa":
            print("IA: Diseñando cocina elegante francesa.")

        else:
            print("IA: Cocina internacional personalizada.")

        print("Diseño del chef terminado.")

    # OPCIÓN 2
    elif opcion == "2":

        print("\n--- MENÚ GOURMET ---")

        restaurante = input("¿Tipo de restaurante? ")

        if restaurante == "lujoso":
            print("IA: Diseñando menú exclusivo gourmet.")

        elif restaurante == "familiar":
            print("IA: Creando menú tradicional y variado.")

        else:
            print("IA: Creando restaurante personalizado.")

        menu = input("¿Menú insignia? ")

        if menu == "mariscos":
            print("IA: Agregando platillos del mar premium.")

        elif menu == "carnes":
            print("IA: Diseñando cortes finos especiales.")

        else:
            print("IA: Creando menú creativo único.")

        print("Menú gourmet terminado.")

    # OPCIÓN 3
    elif opcion == "3":

        print("\n--- BRIGADA DE COCINA ---")

        brigada = input("¿Brigada pequeña o grande? ")

        if brigada == "pequeña":
            print("IA: Organizando equipo compacto y eficiente.")

        else:
            print("IA: Coordinando chefs y ayudantes de cocina.")

        anos = input("¿Años en cocina? ")

        if anos == "1":
            print("IA: Chef principiante con mucho potencial.")

        elif anos == "5":
            print("IA: Chef experimentado en cocina profesional.")

        elif anos == "10":
            print("IA: Chef experto con gran trayectoria.")

        else:
            print("IA: Experiencia culinaria personalizada.")

        print("Brigada organizada.")

    # OPCIÓN 4
    elif opcion == "4":

        print("\nGracias por usar IA Chef Michelin.")
        print("Programa finalizado.")
        break

    # OPCIÓN INVÁLIDA
    else:
        print("\nOpción no válida.")
```
