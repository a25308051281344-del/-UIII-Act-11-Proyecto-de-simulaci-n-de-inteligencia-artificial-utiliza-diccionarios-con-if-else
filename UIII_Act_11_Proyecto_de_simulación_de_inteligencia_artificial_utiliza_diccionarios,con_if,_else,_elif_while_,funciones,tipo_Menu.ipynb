#### 29.Chef: Estrellas Michelin, especialidad culinaria, restaurante, brigada a cargo, menú insignia y años en cocina.

```python
print("====================================================")
print("               IA CHEF MICHELIN")
print("====================================================")

while True:

    print("\n================ MENÚ PRINCIPAL ================")
    print("1. Crear chef Michelin")
    print("2. Diseñar menú gourmet")
    print("3. Organizar brigada de cocina")
    print("4. Salir")

    opcion = input("\nSelecciona una opción: ")

    # OPCIÓN 1
    if opcion == "1":

        print("\n=============== CHEF MICHELIN ===============")

        nombre = input("Nombre del chef: ")
        estrellas = input("Estrellas Michelin: ")
        especialidad = input("Especialidad culinaria: ")

        print("\n====================================================")
        print("| NOMBRE        | ESTRELLAS | ESPECIALIDAD        |")
        print("====================================================")
        print("|", nombre.ljust(13), "|", estrellas.ljust(10), "|", especialidad.ljust(20), "|")
        print("====================================================")

        if estrellas == "1":
            print("IA: Chef reconocido a nivel local.")

        elif estrellas == "2":
            print("IA: Chef con cocina de alta calidad.")

        elif estrellas == "3":
            print("IA: Chef de clase mundial.")

        else:
            print("IA: Chef en proceso de reconocimiento.")

    # OPCIÓN 2
    elif opcion == "2":

        print("\n=============== MENÚ GOURMET ===============")

        restaurante = input("Nombre del restaurante: ")
        menu = input("Menú insignia: ")
        tipo = input("Tipo de restaurante: ")

        print("\n==============================================================")
        print("| RESTAURANTE         | MENÚ INSIGNIA      | TIPO           |")
        print("==============================================================")
        print("|", restaurante.ljust(20), "|", menu.ljust(20), "|", tipo.ljust(14), "|")
        print("==============================================================")

        if tipo == "lujoso":
            print("IA: Diseñando menú exclusivo gourmet.")

        elif tipo == "familiar":
            print("IA: Creando menú tradicional y variado.")

        else:
            print("IA: Restaurante personalizado.")

    # OPCIÓN 3
    elif opcion == "3":

        print("\n=============== BRIGADA DE COCINA ===============")

        brigada = input("Brigada pequeña o grande: ")
        anos = input("Años en cocina: ")

        print("\n================================================")
        print("| BRIGADA        | AÑOS EN COCINA             |")
        print("================================================")
        print("|", brigada.ljust(15), "|", anos.ljust(27), "|")
        print("================================================")

        if brigada == "pequeña":
            print("IA: Organizando equipo compacto y eficiente.")

        else:
            print("IA: Coordinando chefs y ayudantes.")

        if anos == "1":
            print("IA: Chef principiante.")

        elif anos == "5":
            print("IA: Chef experimentado.")

        elif anos == "10":
            print("IA: Chef experto.")

        else:
            print("IA: Experiencia culinaria personalizada.")

    # OPCIÓN 4
    elif opcion == "4":

        print("\nGracias por usar IA Chef Michelin.")
        print("Programa finalizado.")
        break

    # OPCIÓN INVÁLIDA
    else:
        print("\nOpción no válida.")
```
